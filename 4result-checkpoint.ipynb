```python
import numpy as np
import pandas as pd
import joblib
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib
import warnings
warnings.filterwarnings('ignore')

# ==================== 全局绘图样式 ====================
plt.style.use('seaborn-v0_8-darkgrid')
matplotlib.rcParams['font.family'] = 'sans-serif'
matplotlib.rcParams['font.sans-serif'] = ['Arial', 'DejaVu Sans']
matplotlib.rcParams['axes.labelsize'] = 14
matplotlib.rcParams['axes.titlesize'] = 16
matplotlib.rcParams['legend.fontsize'] = 12
matplotlib.rcParams['xtick.labelsize'] = 12
matplotlib.rcParams['ytick.labelsize'] = 12
matplotlib.rcParams['axes.linewidth'] = 1.2
matplotlib.rcParams['legend.framealpha'] = 0.9
matplotlib.rcParams['legend.edgecolor'] = 'black'
matplotlib.rcParams['legend.fancybox'] = True

# ==================== 1. RC 模拟器 ====================
class TemperatureSimulator:
    """使用 RC 参数和微分方程进行室内温度模拟"""
    def __init__(self, R_ie=0.209, R_im=0.035, C_i=31431, C_m=4064747):
        self.R_ie = R_ie
        self.R_im = R_im
        self.C_i = C_i
        self.C_m = C_m
        self.T_r = None
        self.T_m = None
        self.start_time = None

    def set_initial_conditions(self, start_time_dt, indoor_temp):
        self.start_time = start_time_dt
        self.T_r = indoor_temp
        self.T_m = indoor_temp

    def simulate_step(self, T_out, dt_sec=300):
        dT_r = ((T_out - self.T_r) / self.R_ie + (self.T_m - self.T_r) / self.R_im) / self.C_i
        dT_m = ((self.T_r - self.T_m) / self.R_im) / self.C_m
        self.T_r += dT_r * dt_sec
        self.T_m += dT_m * dt_sec
        return self.T_r

    def simulate_until(self, target_time_dt, outdoor_temp_func, dt_sec=300):
        if self.start_time is None:
            raise ValueError("请先调用 set_initial_conditions")
        T_r = self.T_r
        T_m = self.T_m
        current_time = self.start_time
        while current_time < target_time_dt:
            next_time = current_time + timedelta(seconds=dt_sec)
            if next_time > target_time_dt:
                dt_remain = (target_time_dt - current_time).total_seconds()
                if dt_remain <= 0:
                    break
                T_out = outdoor_temp_func(current_time)
                dT_r = ((T_out - T_r) / self.R_ie + (T_m - T_r) / self.R_im) / self.C_i
                dT_m = ((T_r - T_m) / self.R_im) / self.C_m
                T_r += dT_r * dt_remain
                T_m += dT_m * dt_remain
                break
            else:
                T_out = outdoor_temp_func(current_time)
                dT_r = ((T_out - T_r) / self.R_ie + (T_m - T_r) / self.R_im) / self.C_i
                dT_m = ((T_r - T_m) / self.R_im) / self.C_m
                T_r += dT_r * dt_sec
                T_m += dT_m * dt_sec
            current_time = next_time
        return T_r

# ==================== 2. 室外温度模型（基于关机时刻和次日9点实测温度）====================
def create_outdoor_temp_function(shutdown_outdoor_temp, next_day_9am_outdoor):
    """
    返回一个函数，参数为 hours_from_shutdown（从17:00开始的小时数），返回室外温度。
    采用分段线性：
        - 17:00 (0h) -> shutdown_outdoor_temp
        - 23:00 (6h) -> min_temp = next_day_9am_outdoor - 2   (夜间最低)
        - 03:00 (10h) -> min_temp (维持)
        - 09:00 (16h) -> next_day_9am_outdoor
    """
    min_temp = next_day_9am_outdoor - 2.0

    def outdoor_temp_at_time(hours_from_shutdown):
        if hours_from_shutdown < 0:
            return shutdown_outdoor_temp
        elif hours_from_shutdown < 6:      # 17:00 - 23:00
            ratio = hours_from_shutdown / 6.0
            return shutdown_outdoor_temp + (min_temp - shutdown_outdoor_temp) * ratio
        elif hours_from_shutdown < 10:     # 23:00 - 03:00
            return min_temp
        else:                              # 03:00 - 09:00
            ratio = (hours_from_shutdown - 10) / 6.0
            return min_temp + (next_day_9am_outdoor - min_temp) * ratio
    return outdoor_temp_at_time

# ==================== 3. 生成温度曲线（使用 RC 模拟器） ====================
def generate_temperature_curve(simulator, shutdown_time_dt, shutdown_indoor_temp,
                               shutdown_outdoor_temp, next_day_9am_outdoor,
                               hours=16, dt_sec=300):
    simulator.set_initial_conditions(shutdown_time_dt, shutdown_indoor_temp)
    # 创建室外温度函数（基于关机时刻和次日9点实测温度）
    outdoor_func = create_outdoor_temp_function(shutdown_outdoor_temp, next_day_9am_outdoor)

    def outdoor_at_datetime(current_time):
        hours_from_start = (current_time - shutdown_time_dt).total_seconds() / 3600.0
        return outdoor_func(hours_from_start)

    target_time_dt = shutdown_time_dt + timedelta(hours=hours)
    n_steps = int(hours * 3600 / dt_sec)
    times = [shutdown_time_dt]
    temps = [shutdown_indoor_temp]

    for i in range(1, n_steps):
        current_time = shutdown_time_dt + timedelta(seconds=i*dt_sec)
        T_out = outdoor_at_datetime(current_time - timedelta(seconds=dt_sec))
        simulator.simulate_step(T_out, dt_sec)
        times.append(current_time)
        temps.append(simulator.T_r)
    return times, temps

# ==================== 4. 辅助函数 ====================
def load_lgbm_model(model_path):
    try:
        model = joblib.load(model_path)
        print(f"Successfully loaded LightGBM model: {model_path}")
        return model
    except Exception as e:
        print(f"Failed to load LightGBM model: {e}")
        return None

def get_temperature_from_curve(times, temps, target_time_dt):
    if target_time_dt <= times[0]:
        return temps[0]
    if target_time_dt >= times[-1]:
        return temps[-1]
    idx = min(range(len(times)), key=lambda i: abs(times[i] - target_time_dt))
    return temps[idx]

# ==================== 5. 迭代搜索 ====================
def find_optimal_dt(model, times_curve, temps_curve, target_temp, target_time_dt,
                    shutdown_outdoor_temp, next_day_9am_outdoor,
                    max_iter=20, tol=1.0):
    print("\nStarting iterative search for optimal advance time...")
    assumed_dt = 30.0
    best_dt = None
    best_start_time = None
    results = []
    iteration = 0

    # 获取室外温度函数（用于启动时刻的温度计算）
    outdoor_func = create_outdoor_temp_function(shutdown_outdoor_temp, next_day_9am_outdoor)

    while iteration < max_iter:
        start_time_dt = target_time_dt - timedelta(minutes=assumed_dt)
        indoor_at_start = get_temperature_from_curve(times_curve, temps_curve, start_time_dt)

        # 计算启动时刻距离关机时刻的小时数
        shutdown_time_dt = times_curve[0]
        hours_from_shutdown = (start_time_dt - shutdown_time_dt).total_seconds() / 3600.0
        outdoor_at_start = outdoor_func(hours_from_shutdown)

        features = pd.DataFrame({
            'outdoor': [outdoor_at_start],
            'indoor': [indoor_at_start],
            'indoor_minus_target': [indoor_at_start - target_temp],
            'outdoor_minus_target': [outdoor_at_start - target_temp],
            'temp_interaction': [outdoor_at_start * indoor_at_start],
            'outdoor_sq': [outdoor_at_start ** 2],
            'indoor_sq': [indoor_at_start ** 2],
            'temp_diff': [indoor_at_start - outdoor_at_start]
        })
        predicted_dt = model.predict(features)[0]
        error = abs(assumed_dt - predicted_dt)

        results.append({
            'Iteration': iteration + 1,
            'Assumed (min)': round(assumed_dt, 1),
            'Predicted (min)': round(predicted_dt, 1),
            'Error (min)': round(error, 2),
            'Start Time': start_time_dt.strftime('%H:%M'),
            'Outdoor (℃)': round(outdoor_at_start, 1),
            'Indoor (℃)': round(indoor_at_start, 1)
        })

        print(f"Iter {iteration+1:2d}: Assumed {assumed_dt:6.1f} min → Predicted {predicted_dt:6.1f} min, Error {error:.2f} min")
        print(f"         Start {start_time_dt.strftime('%H:%M')} | Outdoor {outdoor_at_start:.1f}℃ | Indoor {indoor_at_start:.1f}℃")

        if error <= tol:
            best_dt = predicted_dt
            best_start_time = start_time_dt
            print(f"✅ Converged! Optimal advance time: {best_dt:.1f} minutes (error ≤ {tol} min)")
            break

        assumed_dt = assumed_dt + 0.5 * (predicted_dt - assumed_dt)
        iteration += 1

    if best_dt is None:
        best_dt = predicted_dt
        best_start_time = target_time_dt - timedelta(minutes=predicted_dt)
        print(f"⚠️ Max iterations ({max_iter}) reached. Using last predicted: {best_dt:.1f} minutes")

    return best_dt, best_start_time, results

# ==================== 6. 主程序 ====================
def main():
    lgbm_model_path = r'C:\Users\TR\Desktop\aircon_time_predictor.pkl'   # 请修改为实际路径
    lgbm_model = load_lgbm_model(lgbm_model_path)
    if lgbm_model is None:
        return

    print("\n" + "="*60)
    print("   Optimal AC Advance Startup Time System (RC Simulation, Any Target Temp)")
    print("="*60)

    shutdown_indoor_temp = float(input("Enter indoor temperature at shutdown (17:00) (°C): "))
    shutdown_outdoor_temp = float(input("Enter outdoor temperature at shutdown (17:00) (°C): "))
    next_day_9am_outdoor = float(input("Enter outdoor temperature at next day 9 AM (°C): "))
    target_temp = float(input("Enter target temperature (°C) (e.g., 23, 24.5, 26): "))

    simulator = TemperatureSimulator(R_ie=0.107, R_im=0.041, C_i=484500, C_m=32800000)

    shutdown_time_dt = datetime.now().replace(hour=17, minute=0, second=0, microsecond=0)

    print("Generating temperature curve using RC simulation with calibrated outdoor profile...")
    times_curve, temps_curve = generate_temperature_curve(
        simulator, shutdown_time_dt, shutdown_indoor_temp,
        shutdown_outdoor_temp, next_day_9am_outdoor,
        hours=16, dt_sec=300
    )
    print(f"Temperature curve generated with {len(times_curve)} points")

    # 急停判断：自然温度是否已达到目标
    natural_temp_at_target = temps_curve[-1]
    print(f"\nNatural indoor temperature at target time (next day 09:00): {natural_temp_at_target:.1f} °C")
    if natural_temp_at_target <= target_temp:
        print("\n" + "="*60)
        print("   ⚠️  EMERGENCY STOP: No need to start the air conditioner.")
        print(f"   Natural temperature ({natural_temp_at_target:.1f}°C) is already ≤ target temperature ({target_temp}°C).")
        print("   The room will stay cool without any cooling. Skipping optimization.")
        print("="*60)
        # 显示温度曲线
        fig1, ax1 = plt.subplots(figsize=(10, 6), dpi=300)
        ax1.plot(times_curve, temps_curve, color='#1f77b4', linewidth=2, label='Indoor temperature')
        ax1.axhline(y=target_temp, color='#d62728', linestyle='--', linewidth=2, label=f'Target {target_temp}°C')
        ax1.scatter(times_curve[0], temps_curve[0], color='#2ca02c', s=80, zorder=5, label=f'Start: {temps_curve[0]:.1f}°C')
        ax1.scatter(times_curve[-1], temps_curve[-1], color='#9467bd', s=80, zorder=5, label=f'End: {temps_curve[-1]:.1f}°C')
        ax1.set_xlabel('Time (hh:mm)', fontsize=14, labelpad=8)
        ax1.set_ylabel('Temperature (°C)', fontsize=14, labelpad=8)
        ax1.set_title('Indoor temperature evolution (no AC needed)', fontsize=16, fontweight='bold', pad=15)
        ax1.grid(True, linestyle='--', alpha=0.4, linewidth=0.8)
        ax1.legend(loc='best', frameon=True, fancybox=True, shadow=True, fontsize=12)
        ax1.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))
        ax1.xaxis.set_major_locator(mdates.HourLocator(interval=2))
        plt.xticks(rotation=0)
        plt.tight_layout()
        plt.savefig('temperature_profile_no_ac.png', dpi=300, bbox_inches='tight')
        plt.show()
        return

    # 需要制冷，继续优化
    print(f"Natural temperature ({natural_temp_at_target:.1f}°C) is above target ({target_temp}°C). AC is required.")
    print("Proceeding to find optimal startup time...")

    # 图1：室内温度曲线（带目标线）
    fig1, ax1 = plt.subplots(figsize=(10, 6), dpi=300)
    ax1.plot(times_curve, temps_curve, color='#1f77b4', linewidth=2, label='Indoor temperature')
    ax1.axhline(y=target_temp, color='#d62728', linestyle='--', linewidth=2, label=f'Target {target_temp}°C')
    start_temp = temps_curve[0]
    end_temp = temps_curve[-1]
    ax1.scatter(times_curve[0], start_temp, color='#2ca02c', s=80, zorder=5, label=f'Start: {start_temp:.1f}°C')
    ax1.scatter(times_curve[-1], end_temp, color='#9467bd', s=80, zorder=5, label=f'End: {end_temp:.1f}°C')
    ax1.set_xlabel('Time', fontsize=14, labelpad=8)
    ax1.set_ylabel('Temperature (°C)', fontsize=14, labelpad=8)
    ax1.set_title('Indoor temperature evolution after HVAC shutdown', fontsize=16, fontweight='bold', pad=15)
    y_min = min(temps_curve) - 1.0
    y_max = min(temps_curve) + 5.0
    ax1.set_ylim(y_min, y_max)
    ax1.grid(True, linestyle='--', alpha=0.4, linewidth=0.8)
    legend = ax1.legend(loc='best', frameon=True, fancybox=True, shadow=True, fontsize=12)
    legend.get_frame().set_alpha(0.9)
    ax1.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))
    ax1.xaxis.set_major_locator(mdates.HourLocator(interval=2))
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig('temperature_profile.png', dpi=300, bbox_inches='tight')
    plt.show()

    target_time_dt = times_curve[-1]  # 次日09:00

    best_dt, best_start_time, results = find_optimal_dt(
        model=lgbm_model,
        times_curve=times_curve,
        temps_curve=temps_curve,
        target_temp=target_temp,
        target_time_dt=target_time_dt,
        shutdown_outdoor_temp=shutdown_outdoor_temp,
        next_day_9am_outdoor=next_day_9am_outdoor,
        max_iter=20,
        tol=1.0
    )

    print("\n" + "="*60)
    print("                    FINAL RESULT")
    print("="*60)
    print(f"  Target temperature:        {target_temp} °C")
    print(f"  Target time:               next day 09:00")
    print(f"  Optimal advance time:      {best_dt:.1f} minutes")
    print(f"  ▶ Recommended startup time: {best_start_time.strftime('%H:%M')}")
    print("="*60)

    if results:
        df_res = pd.DataFrame(results)
        print("\nIteration details:")
        print(df_res.to_string(index=False))

        # 图2：迭代过程
        fig2, (ax_up, ax_low) = plt.subplots(2, 1, figsize=(11, 8), dpi=300, sharex=True)
        fig2.subplots_adjust(hspace=0.15)

        iterations = df_res['Iteration']
        assumed = df_res['Assumed (min)']
        predicted = df_res['Predicted (min)']

        ax_up.plot(iterations, assumed, 'o-', color='#1f77b4', linewidth=2, markersize=8,
                   label='Assumed advance time')
        ax_up.plot(iterations, predicted, 's-', color='#ff7f0e', linewidth=2, markersize=8,
                   label='Predicted advance time')
        ax_up.fill_between(iterations, assumed, predicted, alpha=0.2, color='gray')
        ax_up.set_ylabel('Advance time (minutes)', fontsize=14, labelpad=8)
        ax_up.set_title('Convergence of iterative optimization', fontsize=16, fontweight='bold', pad=15)
        ax_up.grid(True, linestyle='--', alpha=0.4, linewidth=0.8)
        ax_up.legend(loc='best', fontsize=12, frameon=True, fancybox=True, shadow=True)

        errors = df_res['Error (min)']
        bars = ax_low.bar(iterations, errors, width=0.6, color='#2ca02c', alpha=0.8,
                          edgecolor='black', linewidth=0.8, label='Prediction error')
        for bar, err in zip(bars, errors):
            ax_low.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
                        f'{err:.1f}', ha='center', va='bottom', fontsize=11)
        ax_low.axhline(y=1.0, color='#d62728', linestyle='--', linewidth=2, label='Tolerance (1 min)')
        ax_low.set_xlabel('Iteration number', fontsize=14, labelpad=8)
        ax_low.set_ylabel('Prediction error (minutes)', fontsize=14, labelpad=8)
        ax_low.grid(True, linestyle='--', alpha=0.4, axis='y', linewidth=0.8)
        ax_low.legend(loc='best', fontsize=12, frameon=True, fancybox=True, shadow=True)

        ax_low.set_xticks(iterations)
        ax_low.set_xlim(iterations.min()-0.5, iterations.max()+0.5)
        fig2.suptitle('Startup time optimization results', fontsize=18, fontweight='bold', y=0.98)
        plt.tight_layout(rect=[0, 0, 1, 0.96])
        plt.savefig('optimization_process.png', dpi=300, bbox_inches='tight')
        plt.show()

if __name__ == "__main__":
    main()
```
